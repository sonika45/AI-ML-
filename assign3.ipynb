```python
df["EnrollmentDate"] = pd.to_datetime(df["EnrollmentDate"])
```

```python
df.dtypes
```

```python
import matplotlib.pyplot as plt
```

```python

df["Month"] = df["EnrollmentDate"].dt.month_name()
monthly_enrollment = df.groupby("Month")["StudentID"].count()
df["Month"]
print(monthly_enrollment)
plt.plot(monthly_enrollment.index, monthly_enrollment.values)
plt.xticks(rotation=90)

```

```python
Revenue = df.groupby("Revenue")["Month"].count()
plt.plot(Revenue.index,Revenue.values)
```

```python
df["CourseProgress"].mean()
```

```python
average = df.groupby("Month")["CourseProgress"].mean()
plt.plot(average.index,average.values)
plt.xticks(rotation=90)
plt.figure(figsize=(8,5))
```

```python
average = df.groupby("Month")["AttendancePercentage"].mean()
plt.plot(average.index,average.values)
plt.xticks(rotation=90)
```

```python
number = df.groupby("Course")["StudentID"].count()
plt.bar(number.index,number.values)
plt.xticks(rotation=90)
plt.ylabel("Number of Students")
plt.xlabel("Course")
```

```python
average = df.groupby("Course")["FinalScore"].mean()
plt.bar(average.index,average.values)
plt.xlabel("Course")
plt.ylabel("FinalScore")
plt.xticks(rotation=90)
plt.figure(figsize=(8,5))
```

```python
course = (
    df.groupby("Course")["StudentID"]
      .count()
      .sort_values(ascending=False)
      .head(10)
)
plt.barh(course.index, course.values)
plt.xlabel("Student")
plt.ylabel("Course")
```

```python
course = (
    df.groupby("Instructor")["StudentID"]
      .count()
      .sort_values(ascending=False)
      .head(10)
)
plt.barh(course.index, course.values)
plt.xlabel("Student")
plt.ylabel("instructor")
plt.figure(figsize=(8,5))
```

```python
course = (
    df.groupby("FinalScore")["StudentID"]
      .max()
      .sort_values(ascending=False)
      .head(10)
)
plt.figure(figsize=(8,5))

plt.barh(course.index, course.values)
plt.xlabel("Student")
plt.ylabel("FinalScore")
```
