```python
import pandas as pd 
```

```python
df = pd.read_csv("Datasets/dataset.csv")
df
```

```python
import matplotlib.pyplot as plt

course = df.groupby("course")["age"].count()

plt.figure(figsize=(10, 5))
plt.bar(course.index, course.values , color="green", edgecolor="blue")
plt.xlabel("Course")
plt.ylabel("Age Count")
plt.title("Age by Course")
plt.xticks(rotation=45)
plt.show()
#Bargraph
```

```python
colors =[
    "green",
    "blue",
    "orange",
    "purple",
    "brown",
    "yellow",
    "red"
]
plt.pie(course.values, labels=course.index, colors=colors, shadow="true",explode=(0.5,0,0,0,0,0,0), startangle=90)
plt.title("PieChart")

plt.figure(figsize=(12,8))
#piechart
```
